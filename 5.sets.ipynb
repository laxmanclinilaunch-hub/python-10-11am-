- Sets:
   - sets  are unorder collection of unique elements

   - we are construct the NON_empty sets by **{ }**  and for empty_sets by **set()**

   - sets can't have duplicates

   - sets are mutable like  list 



```python
# creating empty sets

empty_set=set()

empty_set
```

```python
type(empty_set)
```

```python
# non-empty set 
a={1,5,9,"ravi","manisha","bhavana"}
a
```
