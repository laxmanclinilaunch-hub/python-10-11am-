```python
# finding duplicates
def f4(x):
    for j in set(x):
        if x.count(j)>1:
            print(j)
```

```python
# finding unique values
def f5(x):
    for j in set(x):
        if x.count(j)==1:
            print(j)
```

```python
#palindrom

def palindrome(strng):
    if strng==strng[::-1]:
        print("palindrome")
    else:
        print("not a palindrome")   
```

```python
a="bhavana"

palindrome(a)
```

```python
b="madam"

palindrome(b)
```

# - Docstring for the function

```python
def odd_even(num):
    """
    this functions checks if a given number is odd or even 
    input: any validnumber / integer 
    output: odd/ even
    created Date:  8-9-2026
    """

    if type(num)==int:
        if num%2==0:
            return f" even_number:  {num}"
        else:
            return f" odd_number: {num}"
    else:
        return "wrong input"    
```

```python
odd_even(6)
```

```python
odd_even(9)
```

```python
print(odd_even.__doc__)
```

```python
def caculate_area(length, width):
   """
    Calculate the area of a rectangle.

    Parameters:
    length (float): The length of the rectangle.
    width (float): The width of the rectangle.

    Returns:
    float: The area of the rectangle.
    """
   
   return length*width

def calculate_perimeter(length , width):
    """
    Calculate the perimeter of a rectangle.

    Parameters:
    length (float): The length of the rectangle.
    width (float): The width of the rectangle.

    Returns:
    float: The perimeter of the rectangle.
    """
    return 2*(length+width)

def print_shape (length, width):
    """
    Print the area and perimeter of a rectangle.

    Parameters:
    length (float): The length of the rectangle.
    width (float): The width of the rectangle.
    """
        

    area = caculate_area(length, width)
    perimeter = calculate_perimeter(length, width)
    print(f"Rectangle with length {length} and width {width}:")
    print(f"Area: {area}")
    print(f"Perimeter: {perimeter}")
```

```python
print_shape(5,2)
```

```python
print_shape(12,2)
```
