- else:

- the else block will be excuted if no exception was raised in the try block

- if exception block was raised in the try block , else block will not excuted

```python
try:
    f= open("files/sample3.txt","r")

except FileNotFoundError:
    print("file not found")

else:
    print(f.read())          
```

```python
try:
    f= open("files/sample4.txt","r")

except FileNotFoundError:
    print("file not found")

else:
    print(f.read())  
```

- finally

thid block will always be excuted irrespective of whether try or an exception was raised or not 





```python
try:
    f= open("files/sample4.txt","r")

except FileNotFoundError:
    print("file not found")

else:
    print(f.read())  

finally:
    print("it always excuted")    
```

custom exception

```python
class age_notvalid(Exception):
    def __init__(self, message="Age is not valid"):
        self.message=message
        super().__init__(self.message)





def validate_age(age):
        if age<0:
            raise age_notvalid("age shloud be positive")    
        elif age>200:
            raise age_notvalid("age must be lessthan 200")
        else:
            print("age is valid")
    


try:
    age=int(input("enter the age: "))
    validate_age(age)

except  age_notvalid as e:
    print(e)

```
