```python
def caculate_area(length, width):
   """
    Calculate the area of a rectangle.

    Parameters:
    length (float): The length of the rectangle.
    width (float): The width of the rectangle.

    Returns:
    float: The area of the rectangle.
    """
   
   return length*width

def calculate_perimeter(length , width):
    """
    Calculate the perimeter of a rectangle.

    Parameters:
    length (float): The length of the rectangle.
    width (float): The width of the rectangle.

    Returns:
    float: The perimeter of the rectangle.
    """
    return 2*(length+width)

def print_shape (length, width):
    """
    Print the area and perimeter of a rectangle.

    Parameters:
    length (float): The length of the rectangle.
    width (float): The width of the rectangle.
    """
        

    area = caculate_area(length, width)
    perimeter = calculate_perimeter(length, width)
    print(f"Rectangle with length {length} and width {width}:")
    print(f"Area: {area}")
    print(f"Perimeter: {perimeter}")
```

```python
print_shape(5,2)
```

```python
print_shape(12,2)
```

- Decarator

A Python decorator is a design pattern that allows you to modify or extend the behavior of a function or method without directly changing its source code

```python
def deco(x):
    def fun():
        print("****************")
        x()
        print("#################")
    return fun
```

```python
@deco
def hello():
    print("hello people")
```

```python
hello()
```

```python
@deco
def hello2():
    print("bhavana is joined on vinayaka festival")
```

```python
hello2()
```

```python
def doctor_only(func):
    def wrapper(role,patient_name):
        if role=="doctor":
            return func(role,patient_name)
        else:
            print("Access Denied : only doctors can view patient records")
    return wrapper
```

```python
@doctor_only
def view_patient_records(role,patient_name):
    print(f'patient record displayed for :{patient_name}')
```

```python
view_patient_records("doctor","raju")
```

```python
view_patient_records("nurse","raju")
```

```python
def add_consultation_fee(func):
    def wrapper(patient,doctor):
        print(f"patient : {patient}")
        print(f"Doctor : {doctor}")
        print("Consulatation fee :500 INR")
        func(patient,doctor)


    return wrapper    
```

```python
@add_consultation_fee
def appointment(patient,doctor):
    print("Appointment booked successfully!")
```

```python
appointment("raju","Dr Manisha")
```
