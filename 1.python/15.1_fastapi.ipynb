# 1. What is FastAPI?

FastAPI is a modern, high-performance web framework for Python used to develop APIs and backend applications.

# 2. What is an API?

API stands for Application Programming Interface. It allows two different software applications to communicate with each other.

# 3. Why is FastAPI used?

FastAPI is used to:

Build REST APIs
Develop backend applications
Connect frontend and backend
Deploy Machine Learning models
Develop microservices


# 4. Main Features of FastAPI


High performance
Easy to learn and use
Automatic data validation
Automatic API documentation
Python type-hint support
Supports asynchronous programming
Suitable for REST API development
Good support for Machine Learning model deployment


# 5. FastAPI and Python

FastAPI is a Python-based framework. It makes use of Python type hints to understand the expected data types and perform validation.

# 6. Pydantic

Pydantic is used by FastAPI for data validation and data parsing.

It checks whether the data received by an API has the correct:

Data type
Structure
Required fields
Values

# 7. Uvicorn

Uvicorn is an ASGI web server commonly used to run FastAPI applications.

FastAPI is the framework, while Uvicorn acts as the server that runs the application.

# 8. REST API

FastAPI is commonly used to create RESTful APIs.

REST APIs generally use HTTP methods such as:

**GET – Retrieve data**

**POST – Create data**

**PUT – Update data**

**DELETE – Delete data**


# 13. Response

A response is the data returned by the server after processing an API request. FastAPI commonly returns responses in JSON format.

# 14. Automatic Documentation

FastAPI automatically generates API documentation.

It provides:

Swagger UI
ReDoc

This allows developers to understand and test APIs easily.

# 15. Asynchronous Programming

FastAPI supports asynchronous programming using Python's **async and await**

This is useful for applications that perform many I/O operations, such as database and network requests.
