# - writing of the file 

```python
f=open("files/sample.txt" , "w")

f.write("Hello python")

f.close()
```

```python
# writing multiple lines 



f=open("files/sample1.txt","w")

f.write("Hello python")
f.write("\nHow are you?")

f.close()
```

```python
# adding extra data into existing file  using append

f=open("files/sample.txt" , 'a')

f.write("\n this is laxman")

f.close()
```

```python
x=["Hello\n" , "Hi\n", "How are you\n","I am fine"]

f=open("files/sample2.txt", "w")

f.writelines(x)

f.close()
```

# - Reading of the files

```python
f=open("files/sample.txt" , "r")

s=f.read()
print(s)
f.close()
```

```python
f=open("files/sample1.txt" , "r")

s=f.read()
print(s)
f.close()
```

```python
f=open("files/sample2.txt" , "r")

s=f.read()
print(s)
f.close()
```

```python
# reading upto n chars

f=open("files/sample2.txt" , "r")
s=f.read(10)
print(s)
f.close()
```

```python
# reading line by line


f=open("files/sample.txt" , "r")
print(f.readlines(),end='')
f.close()
```
