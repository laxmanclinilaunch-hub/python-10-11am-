- Truthy  & Falsy statements


- All values considered to be "Truthy" except following

- None

- False

- 0

- 0.0

- empty str  " "

- empty list  [ ]


- empty tuples ( )


- empty set = set()

- empty dict { }

```python
if []:
    print("what do you say?")
else:
    print("I have nothing to say")    
```

```python
if False:
    print("I have nothing to say")


elif []:
    print("what are you thinking?") 


elif None:
    print("this is falsy statements") 

else:
    print("all above conditions are falsy")
          
```
