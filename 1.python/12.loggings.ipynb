Logging:

it is a crucial aspect of any application

it is used to track the events, errors, total operational information

python built-in logging module offers a flexible framework for emting log messages from python program

```python
import logging

import os
```

```python
log_dir="loggings"

log_file_name="app.log"




os.makedirs(log_dir, exist_ok=True)

log_path=os.path.join(log_dir, log_file_name)
```

```python
logging.basicConfig(filename=log_path, level=logging.INFO)
```

Log levels

- 1. info:

confirmation that things are working expected

```python
logging.info("This is an info message that will be logged to the file")
```

- warning:

An indication that something unexcepted happend or indicating of some problem in the near future

example :
 low diskspace

```python
logging.warning("this is first wartning")
```

- critical:
  
  it is very serious probelm, it effect the continues running of the problem

```python
logging.critical("this is my critical error")
```

- Error:

Due to some serious problem , that software hasnot been able to perform some function

```python
logging.error("this is my first error")
```

- Debug:

detialed infornation , it works when only diagnosing the problem

```python
logging.debug("this is my first debug")
```

- advance logs implementation

```python
LOG_DIR = "loggings"
LOG_FILE_NAME = "app.log"

os.makedirs(LOG_DIR, exist_ok=True)

log_path = os.path.join(LOG_DIR,LOG_FILE_NAME)
```

```python
logging.basicConfig(filename=log_path, level=logging.INFO, format = "[ %(asctime)s] %(name)s - %(levelname)s - %(message)s",force=True)
```

```python
def add(a,b):
    result=a+b
    logging.info(f"Adding {a} + {b}= {result}")
    return result

def subtract(a, b):
    result = a - b
    logging.info(f"Subtracting {a} - {b} = {result}")
    return result

def multiply(a, b):
    result = a * b
    logging.info(f"Multiplying {a} * {b} = {result}")
    return result

def divide(a, b):
    try:
        result = a / b
        logging.debug(f"Dividing {a} / {b} = {result}")
        return result
    except ZeroDivisionError:
        logging.error("Division by zero error")
        return None
    
add(10,15)
subtract(15,10)
multiply(10,20)
divide(20,0)
```
