# - Break , continue , pass

# Break:
  - when specific condition met, break allows to terminate the loop

```python
a="bhavana"

for  i in a:
    if i=="v":
        break
    print(f"this is my break condition : {i}")
    
```

# Continue:
   - when specific condition met, continue allows to skip the loop and control transfered to next statement

```python
a="bhavana"

for  i in a:
    if i=="v":
        continue
    print(f"this is my continue condition : {i}")
    
```
